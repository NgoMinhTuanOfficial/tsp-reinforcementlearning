# TSPLIB Active Search

This notebook runs the repository source directly. A fresh Pointer Network is randomly initialized for this instance and optimized online; no pretrained model or RL pretraining is used. Replace `INSTANCE_PATH` with any supported TSPLIB `.tsp` file.

```python
from pathlib import Path
import sys

REPO_ROOT = Path.cwd()
if not (REPO_ROOT / "src").is_dir():
    REPO_ROOT = REPO_ROOT.parent
sys.path.insert(0, str(REPO_ROOT / "src"))

INSTANCE_PATH = REPO_ROOT / "data" / "sample10.tsp"
print(f"Repository: {REPO_ROOT}")
print(f"Instance: {INSTANCE_PATH}")
```

```python
from tsp_active_search import ActiveSearchConfig, active_search, load_tsplib
from tsp_active_search.plotting import plot_result

instance = load_tsplib(INSTANCE_PATH)
instance.name, instance.dimension, instance.edge_weight_type
```

```python
# Keep the demo short; increase iterations/batch_size to spend more search compute.
config = ActiveSearchConfig(
    iterations=40,
    batch_size=32,
    learning_rate=1e-3,
    seed=7,
)
result = active_search(instance, config)
print(f"Best tour length: {result.best_length:g}")
print("Tour labels:", " -> ".join(instance.node_ids[i] for i in result.best_tour))
```

```python
plot_result(instance, result);
```
